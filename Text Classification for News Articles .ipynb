# AG News Classification

## Overview

This project involves classifying news articles using the AG News dataset. The dataset consists of news articles categorized into different classes. The project demonstrates a text classification pipeline using natural language processing (NLP) and machine learning techniques.

## Dataset

The dataset includes two CSV files:
- **Training Data (`train.csv`)**: Contains news articles used for training the model.
- **Test Data (`test.csv`)**: Contains news articles used for evaluating the model.

### Columns:
- **Class Index**: Numeric index representing the class of the news article.
- **Title**: Title of the news article.
- **Description**: Full text of the news article.

## Objectives

### Data Exploration and Preprocessing:
1. Load the dataset using pandas.
2. Preprocess the text by tokenizing, converting to lowercase, and removing stopwords.
3. Encode the categorical labels into numeric values.

### Model Building and Evaluation:
1. Implement a text classification model using a pipeline that combines TF-IDF vectorization and Multinomial Naive Bayes classification.
2. Train the model on the training dataset.
3. Evaluate the model on the test dataset using accuracy and classification metrics.

### Model Persistence:
1. Save the trained model for future use.

## Methods and Tools

### Libraries:
- **pandas**: For data manipulation.
- **nltk**: For text preprocessing.
- **scikit-learn**: For building and evaluating the classification model.
- **joblib**: For saving the trained model.

### Algorithms:
- **TF-IDF Vectorization**: Converts text into numerical features.
- **Multinomial Naive Bayes**: A probabilistic classifier used for text classification.


```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, accuracy_score
import joblib

# Load data
train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')

# Preprocessing
def preprocess_text(text):
    # Add your text preprocessing steps here
    return text

# Apply preprocessing to the text columns
train_df['Description'] = train_df['Description'].apply(preprocess_text)
test_df['Description'] = test_df['Description'].apply(preprocess_text)

# Feature extraction and model pipeline
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', MultinomialNB())
])

# Encode the labels
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(train_df['Class Index'])
y_test = label_encoder.transform(test_df['Class Index'])

# Fit the model
pipeline.fit(train_df['Description'], y_train)

# Predict
y_pred = pipeline.predict(test_df['Description'])

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

# Convert class_names to a list of strings if necessary
class_names = [str(name) for name in label_encoder.classes_]

# Print classification report with class names
print("Classification Report:\n", classification_report(y_test, y_pred, target_names=class_names))

# Save the model
joblib.dump(pipeline, 'news_classifier_model.pkl')
```
