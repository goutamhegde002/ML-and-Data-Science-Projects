````python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk

# Download necessary NLTK resources
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('maxent_ne_chunker')
nltk.download('words')

# Load the dataset (replace 'ner_dataset.csv' with your actual file)
df = pd.read_csv('ner_dataset.csv')

# Function to perform NER using NLTK
def perform_ner(text):
    tokens = word_tokenize(text)
    pos_tags = pos_tag(tokens)
    chunks = ne_chunk(pos_tags)
    entities = []
    for chunk in chunks:
        if hasattr(chunk, 'label'):
            entities.append((chunk.label(), ' '.join(c[0] for c in chunk)))
    return entities

# Sample text for testing
sample_text = df['text'].iloc[0]
print("Sample Text:", sample_text)
print("Named Entities:", perform_ner(sample_text))
````
