# Diabetes Prediction

### Overview
The Diabetes Prediction project involves using a dataset to predict the onset of diabetes based on various health features such as age, BMI, and blood pressure. This dataset is often used to demonstrate classification techniques in machine learning.

### Dataset
The Diabetes dataset is available in the `scikit-learn` library. It includes features related to health measurements and a target variable indicating whether the patient has diabetes.

### Objectives

#### Data Exploration and Preprocessing:
- Load the Diabetes dataset.
- Inspect the dataset for missing values and basic statistics.
- Preprocess the data by handling missing values, encoding categorical variables (if any), and scaling features.

#### Exploratory Data Analysis (EDA):
- Analyze the distribution of each feature.
- Visualize relationships between features and the target variable using plots such as histograms and scatter plots.
- Assess the correlation between features and the target variable (diabetes).

#### Model Building and Evaluation:
- Build classification models using algorithms like Logistic Regression, Decision Trees, and Random Forest.
- Evaluate model performance using metrics such as Accuracy, Precision, Recall, F1-score, and ROC-AUC.

#### Model Tuning:
- Fine-tune hyperparameters to improve model performance.
- Use techniques like cross-validation to assess model stability and performance.

### Methods and Tools

#### Algorithms:
- **Logistic Regression**: A linear model used for binary classification problems.
- **Decision Trees**: A non-linear model that splits data into decision nodes to make predictions.
- **Random Forest**: An ensemble model that aggregates predictions from multiple decision trees.

#### Libraries:
- `scikit-learn` for data processing, model building, and evaluation.
- `pandas` for data manipulation.
- `matplotlib` and `seaborn` for data visualization.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

# Load the dataset
data = load_diabetes()
X = data.data
y = (data.target > 140).astype(int)  # Binary classification: 1 if target > 140, else 0

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize models
models = {
    'Logistic Regression': LogisticRegression(),
    'Decision Tree': DecisionTreeClassifier(),
    'Random Forest': RandomForestClassifier()
}

# Train and evaluate each model
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    print(f"Model: {name}")
    print(f"Accuracy: {accuracy_score(y_test, y_pred)}")
    print(f"Classification Report:\n{classification_report(y_test, y_pred)}")
    print(f"ROC-AUC Score: {roc_auc_score(y_test, model.predict_proba(X_test)[:,1])}\n")
```
